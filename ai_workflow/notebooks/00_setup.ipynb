```python
#@title 1. 掛載 Google Drive，載入 Core { display-mode: "form" }
ROOT = "/content/drive/MyDrive/AI-Workflow"  #@param {type:"string"}
from google.colab import drive
drive.mount("/content/drive")
import sys
sys.path.insert(0, ROOT)
from controller import notebook as nb
session = nb.connect(ROOT, mount=False)
```

```python
#@title 2. 環境檢查與 ComfyUI 安裝 { display-mode: "form" }
actions = nb.setup(session)
```

```python
#@title 3. 模型檢查（哪些已經在 Google Drive） { display-mode: "form" }
nb.check_models(session)
```

```python
#@title 4. （選用）預先下載模型到 Google Drive { display-mode: "form" }
#@markdown 勾選要先備好的 workflow。一次下載一個檔、驗證後直接存進 Drive，不需要 GPU，也不會生成任何東西。
#@markdown 已經在 Drive 上的檔案會略過。
#@markdown `z-image-basic`（Z-Image Turbo 文字生圖，約 20.7 GB）
fetch_z_image_basic = False  #@param {type:"boolean"}
#@markdown `minimax-h3-basic`（MiniMax H3 首幀生影片（含音訊），約 43.1 GB）
fetch_minimax_h3_basic = False  #@param {type:"boolean"}
saved = nb.prefetch_models(session, {"z-image-basic": fetch_z_image_basic, "minimax-h3-basic": fetch_minimax_h3_basic})
```

```python
#@title 5. 儲存並結束 { display-mode: "form" }
#@markdown 確認這個 session 寫入的東西都已進到 Google Drive（flush）。
#@markdown 存完後釋放 runtime（不再消耗運算單元）。取消勾選的話 runtime 會繼續計費，要自己中斷。
release_runtime = True  #@param {type:"boolean"}
summary = nb.finish(session, release_runtime)
```
